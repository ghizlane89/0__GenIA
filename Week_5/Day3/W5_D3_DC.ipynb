```python
# 1. CHARGER L'ENSEMBLE DE DONNÉES MNIST

from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Chargement des données
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Affichage des dimensions
print("x_train.shape :", x_train.shape)  # (60000, 28, 28)
print("y_train.shape :", y_train.shape)  # (60000,)
print("x_test.shape  :", x_test.shape)   # (10000, 28, 28)
print("y_test.shape  :", y_test.shape)   # (10000,)
```

```python
# 2. PRÉTRAITER LES DONNÉES POUR RÉSEAU NEURONAL ENTIÈREMENT CONNECTÉ

# Aplatir les images 28x28 en vecteurs de 784
x_train_flat = x_train.reshape(-1, 784).astype("float32") / 255
x_test_flat = x_test.reshape(-1, 784).astype("float32") / 255

# Encodage one-hot des étiquettes
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

```python
# 3. CONSTRUIRE ET ENTRAÎNER LE RÉSEAU NEURONAL ENTIÈREMENT CONNECTÉ

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Création du modèle dense
model_dense = Sequential()
model_dense.add(Dense(128, activation='relu', input_shape=(784,)))
model_dense.add(Dense(64, activation='relu'))
model_dense.add(Dense(10, activation='softmax'))

# Compilation du modèle
model_dense.compile(optimizer='adam',
                    loss='categorical_crossentropy',
                    metrics=['accuracy'])

# Entraînement
print("\n=== Entraînement modèle Dense ===")
model_dense.fit(x_train_flat, y_train_cat, epochs=5, batch_size=32, validation_split=0.1)

# Évaluation
loss_dense, acc_dense = model_dense.evaluate(x_test_flat, y_test_cat)
print("Précision du modèle Dense :", acc_dense)
```

####  **Évolution de l’apprentissage sur 5 époques :**

Progression de l'apprentissage sur 5 époques
Le modèle a été entraîné pendant 5 époques, avec les résultats suivants :

Dès la première époque, la précision est déjà élevée, atteignant 87 % sur l'entraînement et plus de 96 % sur la validation.

À chaque époque suivante, la précision continue d'augmenter tandis que la perte diminue progressivement, signe d’un bon apprentissage.

Dès la 3e époque, les performances deviennent très stables, avec des résultats presque constants entre l'entraînement et la validation.

Résultats finaux sur le jeu de test
Précision : 97,66 %

Perte : 0,0951

Interprétation des résultats
Le modèle apprend de façon rapide et efficace, avec une évolution régulière des métriques.

Il parvient à bien généraliser, comme le montre la stabilité de la précision sur les données de validation.

Aucun surapprentissage n’est observé : les écarts entre les performances sur l'entraînement et la validation restent très faibles.

```python

# 4. PRÉTRAITER LES DONNÉES POUR CNN

# Ajouter une dimension canal : (60000, 28, 28, 1)
x_train_cnn = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255
x_test_cnn = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255

# Réutilisation des étiquettes one-hot déjà créées : y_train_cat, y_test_cat
```

```python

# 5. CONSTRUIRE ET ENTRAÎNER LE RÉSEAU CONVOLUTIF

from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten

# Création du modèle CNN
model_cnn = Sequential()
model_cnn.add(Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)))
model_cnn.add(MaxPooling2D(pool_size=(2, 2)))
model_cnn.add(Conv2D(64, kernel_size=(3, 3), activation='relu'))
model_cnn.add(MaxPooling2D(pool_size=(2, 2)))
model_cnn.add(Flatten())
model_cnn.add(Dense(128, activation='relu'))
model_cnn.add(Dense(10, activation='softmax'))

# Compilation
model_cnn.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

# Entraînement
print("\n=== Entraînement modèle CNN ===")
model_cnn.fit(x_train_cnn, y_train_cat, epochs=5, batch_size=32, validation_split=0.1)

# Évaluation
loss_cnn, acc_cnn = model_cnn.evaluate(x_test_cnn, y_test_cat)
print("Précision du modèle CNN :", acc_cnn)
```

#### Résumé de l’apprentissage :

Synthèse de l'entraînement (5 époques)
Le modèle CNN progresse rapidement dès la première époque, atteignant une précision de 89,3 % avec une perte de 0,33. Les performances continuent de s’améliorer à chaque époque :

La précision d’entraînement dépasse les 99 % dès la 4e époque, avec une perte qui chute à 0,0148 à la dernière époque.

Côté validation, les résultats sont très stables autour de 99 %, avec une perte faible, signe de bonne généralisation.

Évaluation finale
Précision sur le jeu de test : 99,10 %

Perte sur le jeu de test : 0,0337

Analyse
Le modèle converge très rapidement, et ce dès la première époque.

Il généralise bien sans montrer de signe de surapprentissage (les performances validation/test sont constantes et proches de celles de l’entraînement).

Grâce à ses couches convolutionnelles, il extrait plus efficacement les caractéristiques spatiales des images (formes, bords…), contrairement au modèle dense.

Comparaison entre les deux architectures
Critère	Modèle Dense	Modèle CNN
Précision test	97,66 %	99,10 %
Perte test	0,0951	0,0337
Convergence	Bonne	Meilleure
Généralisation	Bonne	Excellente
Architecture	Simple	Spécialisée image

Conclusion:
Le CNN s’avère nettement plus adapté aux données visuelles, offrant de meilleures performances en précision, perte et stabilité. C’est le modèle à privilégier pour des tâches de classification d’images comme celle-ci.


```python
# 6. COMPARAISON DES PERFORMANCES

print("\n=== Comparaison des performances ===")
print(f"Modèle Dense - précision : {acc_dense:.4f}")
print(f"Modèle CNN   - précision : {acc_cnn:.4f}")

if acc_cnn > acc_dense:
    print("Le CNN est plus performant que le modèle dense.")
else:
    print("Le modèle dense est plus performant que le CNN.")
```

Résultats
Modèle dense : précision sur le test à 97,66 %

Modèle CNN : précision sur le test à 99,10 %

Analyse
Le modèle CNN dépasse nettement le modèle dense en termes de précision.
L’écart de 1,44 point est significatif, surtout dans un contexte où les performances sont déjà élevées.

Cette différence s’explique par la capacité du CNN à exploiter la structure spatiale des images, grâce aux couches convolutionnelles.
En comparaison, le modèle dense traite chaque pixel de manière isolée, ce qui limite son efficacité pour des images.

Conclusion
Le CNN est clairement plus adapté à la classification d’images comme celles du jeu de données MNIST.
Il offre de meilleures performances globales et capture plus finement les caractéristiques visuelles importantes.
