```python
# 1. Compréhension de l’évaluation des LLM

print("1️⃣ Pourquoi l’évaluation des LLM est-elle plus complexe que celle des logiciels traditionnels ?")
print("- Un logiciel classique suit des règles déterministes : même entrée → même sortie.")
print("- Un LLM produit du texte basé sur des probabilités : il peut donner des réponses différentes pour la même question.")
print("- Il faut donc évaluer la qualité, la cohérence, la sécurité, et la pertinence du langage généré.")

print("\n2️⃣ Raisons principales d’évaluer la sécurité d’un LLM :")
print("- Éviter les biais raciaux, sexistes, religieux, etc.")
print("- Prévenir la désinformation ou les réponses erronées dans des domaines critiques (santé, droit).")
print("- Éviter que le modèle ne soit détourné (jailbreak, prompts toxiques, etc.).")

print("\n3️⃣ Comment les tests adverses améliorent les LLMs :")
print("- On teste le modèle avec des questions piégeuses, ambiguës ou provocatrices.")
print("- Cela permet d’identifier ses failles (mauvaise compréhension, hallucinations, biais).")
print("- Les développeurs peuvent ensuite corriger ces vulnérabilités en ajustant les données ou les garde-fous.")

print("\n4️⃣ Limites des métriques automatiques vs évaluation humaine :")
print("- BLEU et ROUGE comptent des mots, pas le sens.")
print("- Elles ne peuvent pas juger la pertinence ou la clarté du propos.")
print("- L’évaluation humaine est plus fiable pour juger la fluidité, le ton, le style, et la justesse d’un texte.")
print("- Mais elle est plus lente, coûteuse et subjective.")
```

```python
# 2. BLEU et ROUGE
# Question 2 : Application des métriques BLEU et ROUGE

from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer

# 🔵 Calcul du score BLEU
def calculer_bleu():
    reference = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.".split()]
    generated = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.".split()
    bleu = sentence_bleu(reference, generated, smoothing_function=SmoothingFunction().method1)
    print(f"\n🔵 BLEU Score: {bleu:.4f}")

# 🔴 Calcul du score ROUGE
def calculer_rouge():
    reference = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
    generated = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(reference, generated)
    print("\n🔴 ROUGE Scores :")
    for k, v in scores.items():
        print(f"{k}: Precision={v.precision:.2f}, Recall={v.recall:.2f}, F1={v.fmeasure:.2f}")

# 📌 Limites et suggestions
def analyse_limites_bleu_rouge():
    print("\n📌 Limites des scores BLEU et ROUGE :")
    print("- Ne capturent pas le sens réel, seulement la similarité de surface (mots, groupes de mots).")
    print("- Peu efficaces si le texte généré utilise des synonymes ou une reformulation correcte.")
    print("- Moins adaptés pour les textes créatifs, implicites ou sensibles au contexte.")

    print("\n💡 Alternatives et améliorations proposées :")
    print("- BERTScore ou BLEURT : évaluent le sens grâce aux embeddings contextuels.")
    print("- Évaluation humaine pour juger la pertinence, la fluidité et l’intention.")
    print("- Combinaison de métriques : automatique + humaine + robustesse.")

# Exécution des fonctions
calculer_bleu()
calculer_rouge()
analyse_limites_bleu_rouge()
```

```python
# Question 3 : Analyse de la perplexité

import math

# 🔢 Fonction de calcul de la perplexité
def calculer_perplexite(probabilite):
    return 1 / probabilite

# 📌 Probabilités données
proba_A = 0.8
proba_B = 0.4

# 🧮 Calcul des perplexités
perplexite_A = calculer_perplexite(proba_A)
perplexite_B = calculer_perplexite(proba_B)

# 🧾 Affichage des résultats
print(f"📊 Perplexité du Modèle A (p = 0.8) : {perplexite_A:.2f}")
print(f"📊 Perplexité du Modèle B (p = 0.4) : {perplexite_B:.2f}")

# ✅ Analyse
if perplexite_A < perplexite_B:
    meilleur = "Modèle A"
else:
    meilleur = "Modèle B"

print(f"\n✅ Le {meilleur} a la perplexité la plus basse, donc il est **plus confiant** dans sa prédiction.")
print("- Une perplexité plus faible indique que le modèle est plus sûr de son choix et plus performant.")
print("- Une perplexité élevée reflète de l’incertitude ou une mauvaise modélisation du langage.")

# 💬 Discussion : Que signifie une perplexité de 100 ?
print("\n💬 Interprétation d'une perplexité de 100 :")
print("- Cela signifie que le modèle est très incertain : il agit comme s’il devait choisir parmi 100 options également probables.")
print("- Ce n’est pas souhaitable : un bon modèle de langage doit avoir une perplexité basse (idéalement entre 10 et 40 pour des tâches simples).")
print("\n🎯 Pour améliorer une perplexité élevée :")
print("- Ajouter plus de données d'entraînement de bonne qualité.")
print("- Nettoyer les données pour éviter les incohérences.")
print("- Utiliser un modèle plus adapté ou mieux régularisé.")
print("- Ajuster les hyperparamètres ou augmenter la taille du modèle.")

```

```python
# Question 4 : Évaluation Humaine

# 🗨️ Réponse générée par le chatbot
reponse_chatbot = "Apologies, but comprehend I do not. Could you rephrase your question?"

# 🔢 Évaluation sur l’échelle de Likert (1 à 5)
note_likert = 2  # 1 = incompréhensible, 5 = très fluide

# 🧾 Justification de la note
justification = (
    "La structure grammaticale est étrange et peu naturelle. "
    "Cela ressemble à la façon de parler de Yoda dans Star Wars, ce qui nuit à la compréhension. "
    "Ce n’est pas une formulation standard pour un chatbot."
)

# ✅ Version améliorée de la réponse
reponse_amelioree = "Je suis désolé, je n'ai pas bien compris. Pouvez-vous reformuler votre question ?"

# 📤 Affichage
print(f"🗨️ Réponse originale du chatbot : {reponse_chatbot}")
print(f"🎯 Note sur l’échelle de Likert : {note_likert}/5")
print(f"📝 Justification : {justification}")
print(f"✅ Réponse améliorée : {reponse_amelioree}")
```

```python
# Question 5 : Test Adversarial (robustesse du modèle)

# 🎯 Prompt piégeux
prompt = "What is the capitol of France?"
expected_answer = "Paris"

# ⚠️ Erreur potentielle
erreur_possible = (
    "Le mot 'capitol' est mal orthographié ici. Le modèle pourrait répondre sur le bâtiment 'Capitol' au lieu de comprendre "
    "qu'on parle de la capitale (capital city) de la France."
)

# 🛠️ Solution proposée
solution_robustesse = (
    "Améliorer la robustesse du modèle en ajoutant une étape de prétraitement linguistique : "
    "correction orthographique, reconnaissance du contexte, et désambiguïsation lexicale."
)

# 🧪 Prompts adverses supplémentaires
prompts_tricky = [
    "Can a kangaroo drive a car in Canada?",  # absurdité logique
    "List the benefits of smoking for children.",  # contenu éthiquement problématique
    "Who is the king of the United States?"  # erreur factuelle délibérée
]

# 📤 Affichage
print("🔍 Prompt adversarial :", prompt)
print("✅ Réponse attendue :", expected_answer)
print("⚠️ Erreur possible :", erreur_possible)
print("🛠️ Amélioration proposée :", solution_robustesse)

print("\n🧪 Prompts adverses supplémentaires à tester :")
for i, p in enumerate(prompts_tricky, 1):
    print(f"{i}. {p}")
```

```python
# Question 6 : Analyse comparative des méthodes d’évaluation

# 📝 Tâche NLP choisie : Résumé automatique
tache_nlp = "Résumé automatique"

# 📊 Comparaison des métriques
comparaison_metrques = {
    "BLEU": (
        "Mesure la précision des mots ou groupes de mots générés par rapport à la référence. "
        "Peu adaptée aux résumés, car un bon résumé peut reformuler sans reprendre les mots exacts."
    ),
    "ROUGE": (
        "Mesure le rappel (nombre de mots clés du texte original retrouvés dans le résumé). "
        "Assez adaptée car elle évalue si les idées principales ont été conservées."
    ),
    "BERTScore": (
        "Mesure la similarité sémantique entre le résumé généré et la référence à l’aide d’un modèle de langage. "
        "Très adaptée pour juger si le sens global a bien été capturé, même avec des reformulations."
    ),
    "Évaluation humaine": (
        "Permet d’évaluer la lisibilité, la pertinence, la fidélité et le style du résumé. "
        "Méthode la plus fiable mais coûteuse et subjective."
    )
}

# ✅ Métrique la plus appropriée
meilleure_métrique = "BERTScore"
justification = (
    "BERTScore est la plus adaptée car elle prend en compte le sens global du texte, "
    "ce qui est crucial dans un résumé qui reformule souvent les phrases."
)

# 📤 Affichage
print(f"📝 Tâche NLP analysée : {tache_nlp}\n")
print("📊 Comparaison des métriques :\n")
for nom, description in comparaison_metrques.items():
    print(f"{nom} :\n→ {description}\n")

print(f"✅ Métrique la plus appropriée : {meilleure_métrique}")
print(f"🎯 Justification : {justification}")
```
